```python
import requests

def query_model(prompt, model="qwen3.8:27b-mlx"):
    r = requests.post("http://localhost:11434/api/chat", json={
        "model": model,
        "messages": [{"role": "user", "content": prompt}],
        "stream": False,
    })
    return r.json()["message"]["content"]

print(query_model("What do llamas eat?"))
```
